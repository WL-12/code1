```python
def chazhao(nums,targe):
    xiabiao = {}
    for i,num in enumerate(nums):
        if targe -num in xiabiao:
            return [xiabiao[targe - num],i]
        xiabiao[num] = i
    return []
```

```python
chazhao([3,3],6)
```

```python
def zichuan(s):
    xiabiao={}
    i = 0
    max_len =0
    for a, b in enumerate(s):
        if b in xiabiao and xiabiao[b] >= i:
            i = xiabiao[b] +1
        xiabiao[b] = a
        max_len = max(max_len,a -i +1)
    return max_len
```

```python
zichuan("pwwkew")
```

```python
def sum(candidates,target):
    candidates.sort()
    zuhe = []
    m = []
    def zhaodao(start , he):
        if he == target:
            zuhe.append(m[:])
            return
        if he >target:
            return 
        for i in range(start, len(candidates)):
            m.append(candidates[i])
            zhaodao(i, he+candidates[i])
            m.pop()
    zhaodao(0,0)
    return zuhe
```

```python
sum([2,3,6,7],6)
```
